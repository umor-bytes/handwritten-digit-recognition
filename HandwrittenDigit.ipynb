```python
# ================================================================
# HANDWRITTEN DIGIT RECOGNIZER
# CNN + INTERACTIVE GRADIO WEB APPLICATION
# Complete Project - Single Google Colab Cell
# ================================================================

# ================================================================
# 1. INSTALL REQUIRED LIBRARIES
# ================================================================

!pip install -q tensorflow numpy matplotlib seaborn scikit-learn pillow gradio

# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import os
import random
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import gradio as gr

from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.models import Sequential

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score
)

from PIL import Image, ImageOps

# ================================================================
# 3. REPRODUCIBILITY
# ================================================================

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)
random.seed(SEED)

print("=" * 75)
print("HANDWRITTEN DIGIT RECOGNIZER")
print("CNN + INTERACTIVE WEB APPLICATION")
print("=" * 75)

print("\nTensorFlow:", tf.__version__)
print("NumPy:", np.__version__)
print("Gradio:", gr.__version__)

# ================================================================
# 4. CHECK GPU
# ================================================================

print("\n" + "=" * 75)
print("HARDWARE CHECK")
print("=" * 75)

gpus = tf.config.list_physical_devices("GPU")

if gpus:
    print("GPU detected!")
    for gpu in gpus:
        print(gpu)
else:
    print("No GPU detected. CPU will be used.")

# ================================================================
# 5. CREATE PROJECT DIRECTORIES
# ================================================================

os.makedirs("results", exist_ok=True)

# ================================================================
# 6. LOAD MNIST DATASET
# ================================================================

print("\n" + "=" * 75)
print("LOADING MNIST DATASET")
print("=" * 75)

(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

print("Training images :", X_train.shape)
print("Training labels :", y_train.shape)
print("Testing images  :", X_test.shape)
print("Testing labels  :", y_test.shape)

# ================================================================
# 7. DISPLAY SAMPLE IMAGES
# ================================================================

plt.figure(figsize=(12, 7))

for i in range(20):

    plt.subplot(4, 5, i + 1)

    plt.imshow(
        X_train[i],
        cmap="gray"
    )

    plt.title(
        f"Digit: {y_train[i]}"
    )

    plt.axis("off")

plt.suptitle(
    "Sample MNIST Handwritten Digits",
    fontsize=16
)

plt.tight_layout()
plt.show()

# ================================================================
# 8. CLASS DISTRIBUTION
# ================================================================

unique, counts = np.unique(
    y_train,
    return_counts=True
)

plt.figure(figsize=(10, 5))

plt.bar(
    unique,
    counts
)

plt.xlabel("Digit")
plt.ylabel("Number of Images")
plt.title(
    "MNIST Training Dataset Class Distribution"
)

plt.xticks(unique)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()

# ================================================================
# 9. PREPROCESSING
# ================================================================

print("\n" + "=" * 75)
print("PREPROCESSING")
print("=" * 75)

# Convert 0-255 → 0-1
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Add channel dimension
# (60000, 28, 28) → (60000, 28, 28, 1)

X_train = np.expand_dims(
    X_train,
    axis=-1
)

X_test = np.expand_dims(
    X_test,
    axis=-1
)

print("Training shape:", X_train.shape)
print("Testing shape :", X_test.shape)

print(
    "Pixel range:",
    X_train.min(),
    "to",
    X_train.max()
)

# ================================================================
# 10. BUILD CNN
# ================================================================

print("\n" + "=" * 75)
print("BUILDING CNN MODEL")
print("=" * 75)

model = Sequential([

    # Input
    layers.Input(
        shape=(28, 28, 1)
    ),

    # Convolutional Block 1
    layers.Conv2D(
        32,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    layers.MaxPooling2D(
        (2, 2)
    ),

    # Convolutional Block 2
    layers.Conv2D(
        64,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    layers.MaxPooling2D(
        (2, 2)
    ),

    # Convolutional Block 3
    layers.Conv2D(
        128,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    # Flatten
    layers.Flatten(),

    # Dense
    layers.Dense(
        128,
        activation="relu"
    ),

    # Dropout
    layers.Dropout(
        0.5
    ),

    # Output
    layers.Dense(
        10,
        activation="softmax"
    )
])

model.summary()

# ================================================================
# 11. COMPILE MODEL
# ================================================================

model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)

# ================================================================
# 12. CALLBACKS
# ================================================================

callbacks = [

    keras.callbacks.EarlyStopping(

        monitor="val_loss",

        patience=3,

        restore_best_weights=True
    ),

    keras.callbacks.ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.5,

        patience=2,

        min_lr=1e-6
    )
]

# ================================================================
# 13. TRAIN MODEL
# ================================================================

print("\n" + "=" * 75)
print("TRAINING CNN")
print("=" * 75)

history = model.fit(

    X_train,

    y_train,

    epochs=15,

    batch_size=128,

    validation_split=0.1,

    callbacks=callbacks,

    verbose=1
)

# ================================================================
# 14. ACCURACY GRAPH
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy",
    linewidth=2
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy",
    linewidth=2
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    "results/accuracy_graph.png",
    dpi=300
)

plt.show()

# ================================================================
# 15. LOSS GRAPH
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["loss"],
    label="Training Loss",
    linewidth=2
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss",
    linewidth=2
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "Training and Validation Loss"
)

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    "results/loss_graph.png",
    dpi=300
)

plt.show()

# ================================================================
# 16. MODEL EVALUATION
# ================================================================

print("\n" + "=" * 75)
print("MODEL EVALUATION")
print("=" * 75)

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print(
    f"\nTest Loss     : {test_loss:.4f}"
)

print(
    f"Test Accuracy : {test_accuracy * 100:.2f}%"
)

# ================================================================
# 17. GENERATE PREDICTIONS
# ================================================================

print("\nGenerating predictions...")

predictions = model.predict(
    X_test,
    verbose=0
)

predicted_classes = np.argmax(
    predictions,
    axis=1
)

# ================================================================
# 18. CLASSIFICATION REPORT
# ================================================================

report = classification_report(
    y_test,
    predicted_classes,
    digits=4
)

print("\n" + "=" * 75)
print("CLASSIFICATION REPORT")
print("=" * 75)

print(report)

# Save report
with open(
    "results/classification_report.txt",
    "w"
) as f:

    f.write(report)

# ================================================================
# 19. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    predicted_classes
)

plt.figure(
    figsize=(10, 8)
)

sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=range(10),

    yticklabels=range(10)
)

plt.xlabel(
    "Predicted Digit"
)

plt.ylabel(
    "Actual Digit"
)

plt.title(
    "Confusion Matrix - MNIST CNN"
)

plt.tight_layout()

plt.savefig(
    "results/confusion_matrix.png",
    dpi=300
)

plt.show()

# ================================================================
# 20. CORRECT / INCORRECT PREDICTIONS
# ================================================================

correct_indices = np.where(
    predicted_classes == y_test
)[0]

incorrect_indices = np.where(
    predicted_classes != y_test
)[0]

print("\n" + "=" * 75)
print("PREDICTION SUMMARY")
print("=" * 75)

print(
    "Total test images     :",
    len(y_test)
)

print(
    "Correct predictions   :",
    len(correct_indices)
)

print(
    "Incorrect predictions :",
    len(incorrect_indices)
)

# ================================================================
# 21. SHOW INCORRECT PREDICTIONS
# ================================================================

if len(incorrect_indices) > 0:

    number_to_show = min(
        20,
        len(incorrect_indices)
    )

    selected_wrong = incorrect_indices[
        :number_to_show
    ]

    plt.figure(
        figsize=(12, 8)
    )

    for i, index in enumerate(
        selected_wrong
    ):

        plt.subplot(
            4,
            5,
            i + 1
        )

        plt.imshow(
            X_test[index].squeeze(),
            cmap="gray"
        )

        confidence = (
            np.max(
                predictions[index]
            ) * 100
        )

        plt.title(
            f"Actual: {y_test[index]}\n"
            f"Predicted: {predicted_classes[index]}\n"
            f"Confidence: {confidence:.1f}%"
        )

        plt.axis("off")

    plt.suptitle(
        "Incorrect Predictions",
        fontsize=16
    )

    plt.tight_layout()

    plt.show()

# ================================================================
# 22. SAVE MODEL
# ================================================================

model_path = (
    "handwritten_digit_recognizer.keras"
)

model.save(
    model_path
)

print(
    "\nModel saved:",
    model_path
)

# ================================================================
# 23. SAVE PROJECT SUMMARY
# ================================================================

summary = f"""
HANDWRITTEN DIGIT RECOGNIZER
============================

Dataset:
MNIST

Training Samples:
{len(X_train)}

Testing Samples:
{len(X_test)}

Image Size:
28 x 28

Number of Classes:
10

Model:
Convolutional Neural Network

Optimizer:
Adam

Loss Function:
Sparse Categorical Crossentropy

Batch Size:
128

Maximum Epochs:
15

Test Loss:
{test_loss:.4f}

Test Accuracy:
{test_accuracy * 100:.2f}%

Correct Predictions:
{len(correct_indices)}

Incorrect Predictions:
{len(incorrect_indices)}
"""

with open(
    "results/project_summary.txt",
    "w"
) as f:

    f.write(summary)

# ================================================================
# 24. FUNCTION TO PREPROCESS USER IMAGE
# ================================================================

def preprocess_user_image(image):

    """
    Converts a user-drawn/uploaded image into
    the 28x28 grayscale format expected by MNIST CNN.
    """

    if image is None:
        return None

    # ------------------------------------------------------------
    # Handle Gradio ImageEditor dictionary
    # ------------------------------------------------------------

    if isinstance(image, dict):

        # Prefer composite image
        if image.get("composite") is not None:
            image = image["composite"]

        # Otherwise background
        elif image.get("background") is not None:
            image = image["background"]

        # Otherwise use first layer
        elif image.get("layers"):
            image = image["layers"][0]

        else:
            return None

    # ------------------------------------------------------------
    # Convert to NumPy
    # ------------------------------------------------------------

    if isinstance(image, Image.Image):

        image = np.array(image)

    image = np.array(image)

    # ------------------------------------------------------------
    # Remove alpha channel
    # ------------------------------------------------------------

    if image.ndim == 3:

        if image.shape[-1] == 4:

            image = image[:, :, :3]

        # RGB → grayscale
        if image.shape[-1] == 3:

            image = np.mean(
                image,
                axis=2
            )

    # ------------------------------------------------------------
    # Normalize to 0-255
    # ------------------------------------------------------------

    image = image.astype(
        np.float32
    )

    if image.max() <= 1.0:

        image *= 255.0

    image = np.clip(
        image,
        0,
        255
    ).astype(
        np.uint8
    )

    # ------------------------------------------------------------
    # MNIST uses black background + white digit
    # ------------------------------------------------------------

    # Determine whether background is bright
    # If background is white, invert it.
    if np.mean(image) > 127:

        image = 255 - image

    # ------------------------------------------------------------
    # Remove unnecessary empty space
    # ------------------------------------------------------------

    coords = np.argwhere(
        image > 20
    )

    if coords.size > 0:

        y_min, x_min = coords.min(
            axis=0
        )

        y_max, x_max = coords.max(
            axis=0
        )

        cropped = image[
            y_min:y_max + 1,
            x_min:x_max + 1
        ]

    else:

        cropped = image

    # ------------------------------------------------------------
    # Resize while maintaining aspect ratio
    # ------------------------------------------------------------

    pil_image = Image.fromarray(
        cropped
    )

    # Make maximum dimension 20 pixels
    # Similar to MNIST digit proportions
    width, height = pil_image.size

    if width == 0 or height == 0:

        return None

    scale = 20.0 / max(
        width,
        height
    )

    new_width = max(
        1,
        int(width * scale)
    )

    new_height = max(
        1,
        int(height * scale)
    )

    pil_image = pil_image.resize(
        (new_width, new_height),
        Image.Resampling.LANCZOS
    )

    # ------------------------------------------------------------
    # Put digit in 28x28 canvas
    # ------------------------------------------------------------

    canvas = Image.new(
        "L",
        (28, 28),
        color=0
    )

    x_offset = (
        28 - new_width
    ) // 2

    y_offset = (
        28 - new_height
    ) // 2

    canvas.paste(
        pil_image,
        (
            x_offset,
            y_offset
        )
    )

    # ------------------------------------------------------------
    # Normalize
    # ------------------------------------------------------------

    processed = np.array(
        canvas
    ).astype(
        "float32"
    ) / 255.0

    # Add channel
    processed = np.expand_dims(
        processed,
        axis=-1
    )

    # Add batch
    processed = np.expand_dims(
        processed,
        axis=0
    )

    return processed


# ================================================================
# 25. PREDICTION FUNCTION
# ================================================================

def predict_digit(image):

    """
    Takes a drawn/uploaded digit and returns:
    - predicted digit
    - confidence
    - probability for all 10 digits
    - processed 28x28 image
    """

    processed = preprocess_user_image(
        image
    )

    if processed is None:

        return (
            "Please draw or upload a digit.",
            "N/A",
            {str(i): 0 for i in range(10)},
            None
        )

    # ------------------------------------------------------------
    # Predict
    # ------------------------------------------------------------

    probabilities = model.predict(
        processed,
        verbose=0
    )[0]

    predicted_digit = int(
        np.argmax(
            probabilities
        )
    )

    confidence = float(
        probabilities[
            predicted_digit
        ] * 100
    )

    # ------------------------------------------------------------
    # Create probability dictionary
    # ------------------------------------------------------------

    probability_dict = {

        str(i): float(
            probabilities[i]
        )

        for i in range(10)
    }

    # ------------------------------------------------------------
    # Convert processed image for display
    # ------------------------------------------------------------

    processed_display = (
        processed[0, :, :, 0] * 255
    ).astype(
        np.uint8
    )

    # ------------------------------------------------------------
    # Output
    # ------------------------------------------------------------

    prediction_text = (
        f"### Predicted Digit: **{predicted_digit}**"
    )

    confidence_text = (
        f"### Confidence: **{confidence:.2f}%**"
    )

    return (
        prediction_text,
        confidence_text,
        probability_dict,
        processed_display
    )


# ================================================================
# 26. BUILD GRADIO WEB APPLICATION
# ================================================================

print("\n" + "=" * 75)
print("BUILDING INTERACTIVE WEB APPLICATION")
print("=" * 75)

custom_css = """

body {
    font-family: Arial, sans-serif;
}

#title {
    text-align: center;
    font-size: 36px;
    font-weight: bold;
    margin-bottom: 5px;
}

#subtitle {
    text-align: center;
    font-size: 17px;
    margin-bottom: 20px;
}

#prediction {
    text-align: center;
    font-size: 25px;
}

"""

with gr.Blocks(
    title="Handwritten Digit Recognizer",
    css=custom_css,
    theme=gr.themes.Soft()
) as demo:

    # ============================================================
    # HEADER
    # ============================================================

    gr.Markdown(
        """
        <div id="title">
        ✍️ Handwritten Digit Recognizer
        </div>

        <div id="subtitle">
        CNN-powered handwritten digit classification
        </div>
        """
    )

    gr.Markdown(
        """
        ### How to use

        **Option 1:** Draw a digit directly on the canvas.

        **Option 2:** Upload an image of a handwritten digit.

        Then click **Predict Digit**.
        """
    )

    # ============================================================
    # INPUT / OUTPUT
    # ============================================================

    with gr.Row():

        # --------------------------------------------------------
        # LEFT SIDE
        # --------------------------------------------------------

        with gr.Column():

            input_image = gr.ImageEditor(

                label="Draw or Upload a Digit",

                type="numpy",

                image_mode="L",

                sources=[
                    "upload",
                    "clipboard"
                ],

                canvas_size=(
                    400,
                    400
                ),

                height=450,

                brush=gr.Brush(
                    default_size=25
                ),

                layers=False
            )

            predict_button = gr.Button(
                "🔍 Predict Digit",
                variant="primary",
                size="lg"
            )

            clear_button = gr.Button(
                "🗑️ Clear"
            )

        # --------------------------------------------------------
        # RIGHT SIDE
        # --------------------------------------------------------

        with gr.Column():

            prediction_output = gr.Markdown(
                "### Predicted Digit: —",
                elem_id="prediction"
            )

            confidence_output = gr.Markdown(
                "### Confidence: —",
                elem_id="prediction"
            )

            probability_output = gr.Label(
                label="Prediction Probabilities",
                num_top_classes=10
            )

            processed_output = gr.Image(
                label="Processed 28 × 28 Image",
                image_mode="L"
            )

    # ============================================================
    # INFORMATION
    # ============================================================

    gr.Markdown(
        """
        ---

        ### 🧠 Model Information

        | Property | Value |
        |---|---|
        | Dataset | MNIST |
        | Model | Convolutional Neural Network |
        | Input | 28 × 28 grayscale image |
        | Classes | 10 (digits 0–9) |
        | Optimizer | Adam |
        | Output | Digit + confidence |

        ### 📊 Model Performance

        This application uses the CNN trained on the MNIST dataset
        in this notebook. The test accuracy is shown below.
        """
    )

    gr.Markdown(
        f"""
        **Test Accuracy: {test_accuracy * 100:.2f}%**
        """
    )

    gr.Markdown(
        """
        ### About the Project

        This project demonstrates how a Convolutional Neural Network
        can learn visual features from handwritten digits and classify
        them into ten categories.

        The application performs image preprocessing, digit
        classification, confidence estimation, and probability
        visualization in real time.
        """
    )

    # ============================================================
    # BUTTON EVENTS
    # ============================================================

    predict_button.click(

        fn=predict_digit,

        inputs=input_image,

        outputs=[
            prediction_output,
            confidence_output,
            probability_output,
            processed_output
        ]
    )

    # Clear button
    clear_button.click(

        fn=lambda: (
            None,
            "### Predicted Digit: —",
            "### Confidence: —",
            {str(i): 0 for i in range(10)},
            None
        ),

        inputs=None,

        outputs=[
            input_image,
            prediction_output,
            confidence_output,
            probability_output,
            processed_output
        ]
    )

# ================================================================
# 27. FINAL PROJECT SUMMARY
# ================================================================

print("\n" + "=" * 75)
print("PROJECT READY")
print("=" * 75)

print(f"""
Dataset             : MNIST
Model               : CNN
Test Accuracy       : {test_accuracy * 100:.2f}%
Test Loss           : {test_loss:.4f}

Correct Predictions : {len(correct_indices)}
Wrong Predictions   : {len(incorrect_indices)}

Model File:
handwritten_digit_recognizer.keras

Results:
results/accuracy_graph.png
results/loss_graph.png
results/confusion_matrix.png
results/classification_report.txt
results/project_summary.txt

Interactive App:
READY
""")

print("=" * 75)
print("LAUNCHING GRADIO APPLICATION...")
print("=" * 75)

# ================================================================
# 28. LAUNCH PUBLIC GRADIO APP
# ================================================================

demo.launch(
    share=True,
    debug=True
)
```
